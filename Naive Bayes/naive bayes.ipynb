```python
import pandas as pd
titanic = pd.read_csv(r'C:\Users\USER\Downloads\Telegram Desktop\titanic.csv')
```

```python
titanic.head()
```

```python
df=titanic.drop(['PassengerId','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis = 'columns')
```

```python
target = titanic.Survived
```

```python
target.head()
```

```python
inputs = df.drop(['Survived'],axis ="columns")
```

```python
inputs.head()
```

```python
dummies = pd.get_dummies(inputs.Sex)
```

```python
dummies.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le_sex =  LabelEncoder()
```

```python
inputs['Sex']=le_sex.fit_transform(inputs['Sex'])
```

```python
inputs.head()
```

```python
inputs.head()
```

```python
inputs.columns[inputs.isna().any()]
```

```python
inputs['Age'] = inputs.Age.fillna(inputs.Age.mean())
```

```python
inputs.columns.isna().any()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(inputs,target,test_size=0.2)
```

```python
from sklearn.naive_bayes import GaussianNB
model =GaussianNB()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
model.predict(x_test[:10])
```

```python
model.predict_proba(x_test[:10])
```
